## For loops

`for i in range(3):` runs the indented body once for each of `0`, `1` and `2`. At each iteration, `i` receives the next value. More generally, for a non-negative integer `n`, `range(n)` supplies the integers from `0` up to, but **not including**, `n`.

The colon `:` introduces the loop body, and **indentation** determines which statements belong to it. Python does not use curly braces or `begin`/`end` to mark these blocks. Here, `print(i)` displays the current value of `i`.


````python
for i in range(3):
    print(i)  # Inside the loop: runs once for each of 0, 1 and 2.
````

## Functions

`def f(x, y):` **defines** a function named `f`. The names `x` and `y` are its **parameters**: they receive values when we call the function. They do not need values before the function is defined.

The indented body runs when we **call** the function, not when we define it. In `f(1, 2)`, the **arguments** `1` and `2` supply the values of `x` and `y`, respectively.

In `a = x + 1`, `=` means assignment: compute the right-hand side, then assign the result to `a`. This `a` is **local** to the function, separate from a variable with the same name outside it.

`return` ends the call and sends the result back to where the function was called.


````python
def f(x, y):
    a = x + 1       # A local name used during this call.
    return a + y    # Send the computed value back; this does not print it.

f(1, 2)             # The notebook displays the returned value: 4.
````

The call `f(1, 2)` returns `4`. **Returning a value is not the same as printing it.** Here we see `4` because Colab displays the value of the final expression in a code cell, unless that value is `None`.

We could instead save the returned value with `answer = f(1, 2)`, or display it explicitly with `print(f(1, 2))`. A function that finishes without a `return` statement returns `None` rather than a computed result.


## Tuples and lists

Lists and tuples both hold an **ordered collection of elements**, which we can access by position.

A **list** is written with square brackets, as in `["a", "b", "c"]`. A **tuple** is commonly written with parentheses and commas, as in `("a", "b", "c")`. The quoted values are **strings**, meaning text.

In the examples below, `l` is a lowercase letter L, not the number 1.


````python
# Square brackets create this list; the quotes mark each element as a string.
l = ["a", "b", "c"]
l  # The notebook displays the list.
````

````python
# Parentheses and commas create this tuple.
t = ("a", "b", "c")
t
````

### Indexing

We access elements of lists and tuples in the same way. For a sequence with `n` elements, the non-negative indices run from `0` to `n - 1`.

`l[0]` means “read the element at index `0` from `l`”. Notice the difference between creating a list with `[ ... ]` and accessing an existing list with `name[index]`.


````python
l[0]  # The first element: 'a'.
````

````python
t[1]  # The second element: 'b'.
````

We can also index from the end: `-1` is the last element, `-2` is the second-to-last element, and so on.


````python
l[-1]  # The last element: 'c'.
````

````python
t[-2]  # The second-to-last element: 'b'.
````

### Changing elements

Lists are **mutable**: we can replace elements in an existing list. Tuples are **immutable**: we cannot replace their elements.

Compare `l[0]`, which reads an element, with `l[0] = "z"`, which assigns a new value at that position in the list.


````python
l = ["a", "b", "c"]
l[0] = "z"  # Replace the first element of this existing list.
l           # Now ['z', 'b', 'c'].
````

````python
# Intentional error: tuples do not support item assignment.
t = ("a", "b", "c")
# t[0] = "z"  # TypeError. Uncomment this line to try it.
````

### Creating a sorted list versus sorting the existing list

When using an operation on a list, check whether it **changes the existing list** or **returns a new one**. These are different behaviours.

`sorted(a)` returns a **new sorted list**, leaving `a` unchanged. By contrast, `a.sort()` sorts the **existing list in place**: it changes that list rather than making a new one. It returns `None`, not the sorted list.

`sort` is a **method**: a function called through an object. In `a.sort()`, the dot selects the method on the list `a`, and the parentheses call it. We will define our own methods in the next notebook.


````python
a = [4, 2, 3, 1]
# Call the built-in function sorted; keep its returned list under a new name.
new_sorted_list = sorted(a)
print(a)                # [4, 2, 3, 1]: unchanged.
print(new_sorted_list)   # [1, 2, 3, 4]: the new list.
````

````python
a = [4, 2, 3, 1]
# Call the sort method on this list; it changes a and returns None.
a.sort()
print(a)  # [1, 2, 3, 4]
````

### A worked example: two lists, one graph

`import matplotlib.pyplot as plt` loads the plotting functions and gives them the short name `plt`. For example, `plt.plot(...)` calls its `plot` function.

The two lists in the next cell give matching coordinates. Their first entries specify the first point, their second entries specify the second point, and so on. They must contain the same number of entries. Here the points lie on the line $y=2x+1$.

**Run the whole cell.** The two drawing commands come before `plt.show()`, so the dots and line appear together in one graph.


````python
import matplotlib.pyplot as plt

x_plot = [-2, -1, 0, 1, 2]       # Horizontal coordinates, in increasing order.
y_plot = [-3, -1, 1, 3, 5]       # Matching vertical coordinates: y = 2*x + 1.

plt.plot(x_plot, y_plot)          # Join consecutive points with straight segments.
plt.scatter(x_plot, y_plot)       # Also mark the five individual points.
plt.xlabel("x")                  # Name the horizontal axis.
plt.ylabel("y")                  # Name the vertical axis.
plt.show()                       # Display the graph explicitly (optional in Colab).
````

**What you should see:** five dots on an increasing straight line, passing through $(0,1)$. The leftmost point is $(-2,-3)$ and the rightmost is $(2,5)$.

| Command | What it does |
| --- | --- |
| `plt.plot(x_plot, y_plot)` | Connects consecutive coordinate pairs with line segments, in the order supplied. |
| `plt.scatter(x_plot, y_plot)` | Shows the coordinate pairs as separate dots, without connecting them. |
| `plt.xlabel("x")`, `plt.ylabel("y")` | Add the axis names. The quoted strings are labels, not variables to evaluate. |
| `plt.show()` | Displays the graph. In these Colab examples, put it last in the plotting cell. |
